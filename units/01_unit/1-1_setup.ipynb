# Project Organization

To keep your research organized and ensure your code can actually find your data, we need to be very intentional about where we save our work.

### Pick a Permanent Home
Do not leave your *montagu* folder in your Downloads folder! Move the entire folder to a dedicated place, such as:

Documents/DHSI/montagu

Desktop/montagu

Or any other location that will be a permanent home to your folder

### Respect the Folder Structure
Inside your *montagu* folder, you will see specific sub-folders like data, units, and img.

Never move these folders.

Our code uses Relative Paths (e.g., ../../data/montagu_2019.txt) to find information. If you move a notebook out of its folder, the code will "break" because it can no longer find the data.

### The "File Explorer" in VS Code
On the left-hand side of your VS Code window, you have a File Explorer.

Use this to see exactly where you are working.

If you see a file sitting alone outside of the montagu folder, it’s in the wrong place!

You can drag and drop files within this side panel to move them into the correct sub-folders.


# Conda Environment:

### 1. Navigate to the course folder

Open VS Code and use `File > Open Folder` to select the *montagu* folder you just cloned.

### 2. Create the Conda environment

Open a **terminal** in VS Code, navigate to the folder where the `environment.yml` file is located, and run the following:

```bash
conda env create -f environment.yml`
```

### 3. Activate the Conda environment

```bash
`conda activate montagu-letters`
```

### 4. Make the environment available as a Jupyter kernel

```bash
`python -m ipykernel install --user --name montagu-letters`
```

### 5. Select the kernel:

In the top-right corner of this VS Code window, click `Select Kernel → Python Environments` and choose `montagu-letters`.

### 6. Deactivate the environment when you're done

```bash
`conda deactivate`
```

# Troubleshooting & Staying Up to Date
To ensure you get the latest materials without losing your own notes, we use the following workflow:

### The "Duplicate" Rule
During the course, as we are running the course notebooks, you will create duplicates of the notebooks and rename them (e.g., *1-1_setup.ipynb* -> *1-1_setup_NOTES.ipynb*)

This way, your renamed files will become your personal space for experiments, notes, and errors. You will always have my original, "clean" code to refer back to. When I add new content to the course, Git can update the original files automatically without clashing with your personal notes.

There are two ways to copy notebooks

1. Copy-Paste & Rename

Right-click a notebook (e.g., 1-1_setup.ipynb) in the VS Code sidebar.

Select Copy, then Paste.

Rename your copy to something like 1-1_setup_working.ipynb.

2. Save As Shortcut

Open the original notebook you want to work on.

Go to the top menu: `File > Save As....`
Type a new name (e.g., 1-1_setup_NOTES.ipynb) and click Save.
VS Code will automatically switch you to the new file so you can start typing right away!

> When using either one of these methods, make sure you save your new _NOTES or _working.ipynb file inside the **same folder as the original**. This ensures that all the links to images and datasets stay intact.

### How to Sync Updates

When I announce that new materials are available, run this command in your terminal:

```bash
git pull origin main
```

### Fixing "Local Changes" Errors

If Git refuses to pull because you accidentally edited an original file, run the following commands to "reset" your folder (be sure to save any important notes elsewhere first!):

```bash
git fetch origin
git reset --hard origin/main
```
